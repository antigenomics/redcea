# 03 Run Clustering Benchmark

Build the benchmark manifests and optionally execute all runs locally. On Slurm, the array jobs consume the generated manifests.

```python
from pathlib import Path
import os
import sys

for candidate in [Path.cwd(), *Path.cwd().parents]:
    if (candidate / "benchmark" / "data_sources.py").exists():
        REPO_ROOT = candidate
        break
else:
    raise RuntimeError("Could not find repository root containing benchmark/data_sources.py")

os.chdir(REPO_ROOT)
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
print(f"Benchmark repo root: {REPO_ROOT}")
```

```python
from benchmark.run_benchmark import build_grid_manifest, build_execution_manifest

grid_manifest = build_grid_manifest(grid_size="small")
execution_manifest = build_execution_manifest("data/processed", grid_size="small")
grid_manifest.to_csv("results/run_metadata/clustering_grid_manifest.tsv", sep="\t", index=False)
execution_manifest.to_csv("results/run_metadata/clustering_execution_manifest.tsv", sep="\t", index=False)
execution_manifest.loc[execution_manifest["dataset_mode"] == "vdjdb"].to_csv("results/run_metadata/clustering_manifest_vdjdb.tsv", sep="\t", index=False)
execution_manifest.loc[execution_manifest["dataset_mode"] == "yfv"].to_csv("results/run_metadata/clustering_manifest_yfv.tsv", sep="\t", index=False)
execution_manifest.head()
```
