```python
import pandas as pd

def _normalize_chain(chain: str) -> str:
    """Accepts alpha/beta or TRA/TRB; returns 'alpha' or 'beta'."""
    chain = chain.strip().upper()
    if chain in {"ALPHA", "TRA", "TRAV", "TCRA"}:
        return "alpha"
    if chain in {"BETA", "TRB", "TRBV", "TCRB"}:
        return "beta"
    raise ValueError(f"Unknown chain: {chain!r}. Use alpha/beta or TRA/TRB.")


def get_epitope_specific_clonotypes(
    df: pd.DataFrame,
    epitope: str,
    chain: str = "beta",
):
    """
    Epitope-specific clonotypes:
      • clonotype is associated ONLY with the given epitope across the dataset
        (considering rows where antigen_epitope is not null)

    Returns a DataFrame with clonotype columns and:
      - epitope
      - n_studies (unique reference_id count within the epitope rows)
    """
    chain = _normalize_chain(chain)

    clonotype_cols = [f"cdr3_{chain}", f"v_{chain}", f"j_{chain}"]

    df_annot = df[df["antigen_epitope"].notna()].copy()
    df_annot = df_annot.dropna(subset=clonotype_cols)

    # epitope set per clonotype
    epi_sets = (
        df_annot
        .groupby(clonotype_cols)["antigen_epitope"]
        .agg(lambda x: set(x))
    )

    specific_idx = epi_sets[epi_sets == {epitope}].index

    df_target = df_annot[df_annot["antigen_epitope"] == epitope].copy()
    df_target = df_target.set_index(clonotype_cols).loc[specific_idx].reset_index()

    n_studies = (
        df_target
        .groupby(clonotype_cols)["reference_id"]
        .nunique()
        .rename("n_studies")
        .reset_index()
    )

    n_studies["epitope"] = epitope
    return n_studies


def get_public_epitope_specific_clonotypes(
    df: pd.DataFrame,
    epitope: str,
    chain: str = "beta",
    min_studies: int = 2,
):
    """
    Public + epitope-specific clonotypes:
      • epitope-specific (only this epitope across dataset)
      • present in >= min_studies studies (unique reference_id)

    Returns a DataFrame with clonotype columns and:
      - epitope
      - n_studies
    """
    specific = get_epitope_specific_clonotypes(df, epitope=epitope, chain=chain)
    public = specific[specific["n_studies"] >= int(min_studies)].copy()
    return public


def topN_specific_and_public(
    df: pd.DataFrame,
    chain: str = "beta",
    min_studies: int = 2,
    n_epi: int = 20,
):
    """
    For top-N epitopes (by #unique clonotypes) compute:
      - specific clonotypes
      - public+specific clonotypes (>= min_studies studies)

    Returns: (specific_df, public_df, top_epitopes_series)
    """
    chain_norm = _normalize_chain(chain)
    cols = [f"cdr3_{chain_norm}", f"v_{chain_norm}", f"j_{chain_norm}"]

    uniq = (
        df[df["antigen_epitope"].notna()]
        .dropna(subset=cols)
        .drop_duplicates(subset=["antigen_epitope"] + cols)
        .groupby("antigen_epitope")
        .size()
        .sort_values(ascending=False)
    )

    top_epitopes = uniq.head(n_epi).index.tolist()

    specific_parts = [get_epitope_specific_clonotypes(df, ep, chain_norm) for ep in top_epitopes]
    public_parts = [get_public_epitope_specific_clonotypes(df, ep, chain_norm, min_studies) for ep in top_epitopes]

    specific_df = pd.concat(specific_parts, ignore_index=True) if specific_parts else pd.DataFrame()
    public_df = pd.concat(public_parts, ignore_index=True) if public_parts else pd.DataFrame()

    # add chain label for downstream merges / saving
    specific_df["chain"] = chain_norm
    public_df["chain"] = chain_norm

    return specific_df, public_df, uniq.head(n_epi)
```

```python
def top10_public_specific(df, chain="beta", min_studies=2, n_epi=20):

    cols = [f"cdr3_{chain}", f"v_{chain}", f"j_{chain}"]
    if chain == "beta" and f"d_{chain}" in df.columns:
        cols.append(f"d_{chain}")

    # --- считаем уникальные клонотипы на эпитоп ---
    uniq = (
        df[df["antigen_epitope"].notna()]
        .drop_duplicates(subset=["antigen_epitope"] + cols)
        .groupby("antigen_epitope")
        .size()
        .sort_values(ascending=False)
    )

    top_epitopes = uniq.head(n_epi).index

    # --- собираем результаты ---
    parts = [
        get_public_epitope_specific_clonotypes(df, ep, chain, min_studies)
        for ep in top_epitopes
    ]

    return pd.concat(parts, ignore_index=True), uniq.head(n_epi)
```

```python
# beta: specific + public
specific_trb, public_trb, top10_trb = topN_specific_and_public(df, chain="TRB", min_studies=2, n_epi=20)

print("Top epitopes (TRB) by #unique clonotypes:")
display(top10_trb.rename("n_unique_clonotypes").to_frame())

print("Specific TRB clonotypes (preview):")
display(specific_trb.head())

print("Public+specific TRB clonotypes (preview):")
display(public_trb.head())
```

```python
# alpha: specific + public
specific_tra, public_tra, top10_tra = topN_specific_and_public(df, chain="TRA", min_studies=2, n_epi=20)

print("Top epitopes (TRA) by #unique clonotypes:")
display(top10_tra.rename("n_unique_clonotypes").to_frame())

print("Specific TRA clonotypes (preview):")
display(specific_tra.head())

print("Public+specific TRA clonotypes (preview):")
display(public_tra.head())
```
