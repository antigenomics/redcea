```python
import pandas as pd

def get_public_epitope_specific_clonotypes(
    df: pd.DataFrame,
    epitope: str,
    chain: str = "beta",
    min_studies: int = 2,
):
    """
    Клонотипы, которые:
      • встречаются >= min_studies исследований
      • во всём датасете ассоциированы только с данным эпитопом
    """

    # --- колонки для цепи ---
    cdr3 = f"cdr3_{chain}"
    v = f"v_{chain}"
    j = f"j_{chain}"

    clonotype_cols = [cdr3, v, j]

    # --- только строки с эпитопом ---
    df_annot = df[df["antigen_epitope"].notna()]

    # --- набор эпитопов на клонотип ---
    epi_sets = (
        df_annot
        .groupby(clonotype_cols)["antigen_epitope"]
        .agg(lambda x: set(x))
    )

    # --- оставляем только те, где эпитоп один и он нужный ---
    specific = epi_sets[epi_sets == {epitope}].index
    
    # --- строки с этим эпитопом ---
    df_target = df[df["antigen_epitope"] == epitope]

    # --- число исследований ---
    study_counts = (
        df_target
        .set_index(clonotype_cols)
        .loc[specific]
        .reset_index()
        .groupby(clonotype_cols)["reference_id"]
        .nunique()
    )

    public = study_counts[study_counts >= min_studies].reset_index()
    public['epitope'] = epitope

    return public
```

```python
def top10_public_specific(df, chain="beta", min_studies=2, n_epi=20):

    cols = [f"cdr3_{chain}", f"v_{chain}", f"j_{chain}"]
    if chain == "beta" and f"d_{chain}" in df.columns:
        cols.append(f"d_{chain}")

    # --- считаем уникальные клонотипы на эпитоп ---
    uniq = (
        df[df["antigen_epitope"].notna()]
        .drop_duplicates(subset=["antigen_epitope"] + cols)
        .groupby("antigen_epitope")
        .size()
        .sort_values(ascending=False)
    )

    top_epitopes = uniq.head(n_epi).index

    # --- собираем результаты ---
    parts = [
        get_public_epitope_specific_clonotypes(df, ep, chain, min_studies)
        for ep in top_epitopes
    ]

    return pd.concat(parts, ignore_index=True), uniq.head(n_epi)
```

```python
# запуск
clono_df, top10 = top10_public_specific(df, chain="beta", min_studies=2)

print("Top-10 epitopes by #unique clonotypes:")
display(top10.rename("n_unique_clonotypes").to_frame())


# --- считаем общее число клонотипов на эпитоп ---
chain = "beta"
cols = [f"cdr3_{chain}", f"v_{chain}", f"j_{chain}"]
if chain == "beta" and f"d_{chain}" in df.columns:
    cols.append(f"d_{chain}")

total_clono = (
    df[df["antigen_epitope"].notna()]
    .drop_duplicates(subset=["antigen_epitope"] + cols)
    .groupby("antigen_epitope")
    .size()
    .rename("n_total_clonotypes")
)

public_clono = (
    clono_df.groupby("epitope")
    .size()
    .rename("n_public_specific_clonotypes")
)

summary = (
    pd.concat([total_clono, public_clono], axis=1)
    .fillna(0)
    .astype(int)
)

summary["fraction_public_specific"] = (
    summary["n_public_specific_clonotypes"] /
    summary["n_total_clonotypes"]
)

summary = summary.loc[top10.index].sort_values(
    "n_public_specific_clonotypes", ascending=False
)

print("\nPublic+specific clonotypes summary:")
display(summary)


# --- топ клонотипов на эпитоп ---
K = 15
for ep in top10.index:
    sub = clono_df[clono_df["epitope"] == ep].copy()

    print(
        f"\n=== {ep} | public+specific: {len(sub)} "
        f"/ total: {summary.loc[ep, 'n_total_clonotypes']} ==="
    )

    if len(sub) == 0:
        continue

    display(sub.head(K))
```
