```python
import pandas as pd

alleles_of_interest = ["HLA-A*01",
                       "HLA-A*24", "HLA-A*11","HLA-A*02", "HLA-A*03",]
interest_set = set(alleles_of_interest)

# превратим строку с аллелями в множество
def parse_hla(x: str) -> set:
    if pd.isna(x):
        return set()
    return set(a.strip() for a in str(x).split(",") if a.strip())

meta_now = meta[meta.cmv == '-'].copy()
meta_now["hla_set"] = meta_now["hla"].apply(parse_hla)

dfs = []
for allele in alleles_of_interest:
    # условие: образец содержит нужную аллель И больше НИКАКИХ других из интересующего списка
    def keep_row(s: set) -> bool:
        present_interest = s & interest_set
        return (allele in s) and (len(present_interest) == 1)

    subset = meta_now[meta_now["hla_set"].apply(keep_row)].copy()

    # берём до 5 (случайно; можно заменить на сортировку для детерминизма)
    k_samples = 7 if '24' in allele else (6 if '03' in allele else 5)
    print(k_samples)
    subset = subset.sample(n=min(k_samples, len(subset)), random_state=42)

    subset["allele"] = allele
    dfs.append(subset)

result = pd.concat(dfs, ignore_index=True).drop(columns=["hla_set"])
result
```
